```python
#Loading The Data
model_data = final_data.copy()
model_data
model_data.info()
```

```python
model_data.columns = model_data.columns.astype(str)
x = model_data.drop(columns = 'Survived')
y = model_data['Survived'].astype(int)
x.info()
```

```python
# HoldOut Method --------> Splitting the data and Cross Validation

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Split the data
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# Identify numerical and categorical columns
numeric_features = x_train.select_dtypes(
    include=['int64', 'float64']
).columns

categorical_features = x_train.select_dtypes(
    include=['str', 'object', 'category']
).columns

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

# Build Logistic Regression model
logistic_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

# Start building the model
logistic_model.fit(x_train, y_train)

# Prediction probabilities
logistic_model_prob = logistic_model.predict_proba(x_test)

# Predictions
logistic_predictions = logistic_model.predict(x_test)

# Display results
print("First 10 Prediction Probabilities:")
print(logistic_model_prob[:10])

print("\nFirst 10 Predictions:")
print(logistic_predictions[:10])
```

```python
# Lets Create an Evaluate Function

def evaluate_model(name, y_true, prediction, probability):

    """Evaluate Function For Metrics"""

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        prediction,
        labels=[0, 1]
    ).ravel()

    return {
        'Model_Name': name,
        'Accuracy': accuracy_score(y_true, prediction),
        'Precision': precision_score(
            y_true,
            prediction,
            zero_division=0
        ),
        'Recall': recall_score(
            y_true,
            prediction,
            zero_division=0
        ),
        'Specificity': tn / (tn + fp),
        'F1-Score': f1_score(
            y_true,
            prediction,
            zero_division=0
        ),
        'ROC-AUC': roc_auc_score(
            y_true,
            probability[:, 1]
        ),
        'TN': tn,
        'FP': fp,
        'TP': tp,
        'FN': fn
    }


# Evaluate Logistic Regression
logistic_metrics = evaluate_model(
    "Logistic Regression",
    y_test,
    logistic_predictions,
    logistic_model_prob
)

logistic_metrics
```

```python
# Lets Create an Evaluate Function

def evaluate_model(name, y_true, prediction, probability):

    """Evaluate Function For Metrics"""

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        prediction,
        labels=[0, 1]
    ).ravel()

    return {
        'Model_Name': name,
        'Accuracy': accuracy_score(
            y_true,
            prediction
        ),
        'Precision': precision_score(
            y_true,
            prediction,
            zero_division=0
        ),
        'Recall': recall_score(
            y_true,
            prediction,
            zero_division=0
        ),
        'Specificity': tn / (tn + fp),
        'F1-Score': f1_score(
            y_true,
            prediction,
            zero_division=0
        ),
        'ROC-AUC': roc_auc_score(
            y_true,
            probability
        ),
        'TN': tn,
        'FP': fp,
        'TP': tp,
        'FN': fn
    }


# We will check for best models
# (Logistic, Decision Tree, Random Forest)

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline

# Identify numerical and categorical columns
numeric_features = x_train.select_dtypes(
    include=['int64', 'float64']
).columns

categorical_features = x_train.select_dtypes(
    include=['str', 'object', 'category']
).columns

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

# Models
models = {

    "Logistic Regression": Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(
            max_iter=1000,
            random_state=42
        ))
    ]),

    "Decision Tree": Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", DecisionTreeClassifier(
            max_depth=5,
            min_samples_leaf=8,
            random_state=42
        ))
    ]),

    "Random Forest": Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", RandomForestClassifier(
            n_estimators=150,
            min_samples_leaf=3,
            random_state=42,
            n_jobs=1
        ))
    ])
}

# Store outputs
test_outputs = {}
metric_rows = []

# Train and evaluate each model
for name, model in models.items():

    model.fit(x_train, y_train)

    # Probability of class 1
    probability = model.predict_proba(x_test)[:, 1]

    # Predictions
    predictions = model.predict(x_test)

    # Store predictions and probabilities
    test_outputs[name] = {
        'probability': probability,
        'predictions': predictions
    }

    # Evaluate model
    metric_rows.append(
        evaluate_model(
            name,
            y_test,
            predictions,
            probability
        )
    )

# Create comparison table
holdout_metrics = pd.DataFrame(metric_rows)

holdout_metrics
```

```python
# Repeated random holdout (10 different stratified splits)

from sklearn.model_selection import StratifiedShuffleSplit
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score

shuffle = StratifiedShuffleSplit(
    n_splits=10,
    test_size=0.20,
    random_state=42
)

random_holdout_rows = []

for name, model in models.items():

    scores = []

    for train_index, test_index in shuffle.split(x, y):

        model.fit(
            x.iloc[train_index],
            y.iloc[train_index]
        )

        probability = model.predict_proba(
            x.iloc[test_index]
        )[:, 1]

        scores.append(
            roc_auc_score(
                y.iloc[test_index],
                probability
            )
        )

    random_holdout_rows.append({
        "Model": name,
        "Mean ROC-AUC": np.mean(scores),
        "Std ROC-AUC": np.std(scores, ddof=1)
    })

random_holdout_results = pd.DataFrame(
    random_holdout_rows
).set_index("Model")

print(random_holdout_results)
```

```python
# Stratified 5-fold cross-validation

from sklearn.model_selection import StratifiedKFold, cross_val_score
import pandas as pd

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_rows = []
cv_auc_scores = {}

for name, model in models.items():

    accuracy_scores = cross_val_score(
        model,
        x,
        y,
        cv=cv,
        scoring="accuracy",
        n_jobs=-1
    )

    f1_scores = cross_val_score(
        model,
        x,
        y,
        cv=cv,
        scoring="f1",
        n_jobs=-1
    )

    auc_scores = cross_val_score(
        model,
        x,
        y,
        cv=cv,
        scoring="roc_auc",
        n_jobs=-1
    )

    cv_auc_scores[name] = auc_scores

    cv_rows.append({
        "Model": name,
        "Accuracy mean": accuracy_scores.mean(),
        "F1 mean": f1_scores.mean(),
        "ROC-AUC mean": auc_scores.mean(),
        "ROC-AUC std": auc_scores.std(ddof=1)
    })

cv_results = (
    pd.DataFrame(cv_rows)
    .set_index("Model")
    .sort_values("ROC-AUC mean", ascending=False)
)

print(cv_results)
```

```python
final_metrics = holdout_metrics[[
    "Accuracy", "Precision", "Recall", "Specificity", "ROC-AUC"
]].sort_values("ROC-AUC", ascending=False)

print("FINAL HOLDOUT METRICS")
display(final_metrics.round(3))

print("\nFINAL CROSS-VALIDATION METRICS")
display(cv_results.round(3))

print(f"\nBest model by mean CV ROC-AUC: {cv_results.index[0]}")
```

```python
# 1. Select the best model using cross-validation ROC-AUC

from sklearn.base import clone

best_model_name = cv_results["ROC-AUC mean"].idxmax()

# Create a fresh copy of the selected model
best_model = clone(models[best_model_name])

# Fit the best model using the complete dataset
best_model.fit(x, y)

print("Best model:", best_model_name)

print(
    "Mean CV ROC-AUC:",
    round(
        cv_results.loc[
            best_model_name,
            "ROC-AUC mean"
        ],
        3
    )
)
```
